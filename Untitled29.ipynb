```python
import numpy as np
import pandas as pd
import torch
from sklearn.metrics import accuracy_score, classification_report
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments
import warnings
warnings.filterwarnings('ignore')

# ─────────────────────────────────────────────
# NOTE ON LABEL STRATEGY
# ─────────────────────────────────────────────
# Dataset 3 contains only label=0 (depression/negative Reddit posts).
# It has no "Normal" class, so we cannot train a binary classifier on
# Dataset 3 alone.  Fix: we augment the training set by pulling the
# "Normal" rows from Dataset 1 (which has never been seen during testing)
# and combining them with Dataset 3.
#
# Final label convention (consistent with original model_2.py):
#   1 = Normal / Non-depressed
#   0 = Depressed / Abnormal
#
# Training set  : Dataset 3 (all label=0) + Dataset 1 "Normal" rows (label=1)
# Test set      : Dataset 1 non-Normal rows  →  Depression=0, everything else=0
#               (i.e. the held-out portion of Dataset 1 mapped to binary)
# ─────────────────────────────────────────────


class SentimentDataset(torch.utils.data.Dataset):
    def __init__(self, texts, labels, tokenizer, max_length=128):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = str(self.texts[idx])
        label = self.labels[idx]

        encoding = self.tokenizer(
            text,
            add_special_tokens=True,
            max_length=self.max_length,
            return_token_type_ids=False,
            padding='max_length',
            truncation=True,
            return_attention_mask=True,
            return_tensors='pt',
        )

        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'labels': torch.tensor(label, dtype=torch.long)
        }


def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    predictions = np.argmax(predictions, axis=1)
    return {'accuracy': accuracy_score(labels, predictions)}


def main():
    print("=" * 60)
    print("Fine-tuning Transformer: Train on Dataset3, Test on Dataset1")
    print("=" * 60)

    # ── Load datasets ──────────────────────────────────────────────
    # Adjust paths if running in Colab, e.g. "/content/Dataset_1.csv"
    df1 = pd.read_csv("/content/Dataset 1.csv")   # columns: statement, status
    df3 = pd.read_csv("/content/dataset3.csv")    # columns: title, content, label

    print("\nDataset 1 columns :", df1.columns.tolist())
    print("Dataset 3 columns :", df3.columns.tolist())

    # ── Prepare Dataset 1 ─────────────────────────────────────────
    # Rename to unified column names
    df1 = df1.rename(columns={'statement': 'text', 'status': 'label'})

    # Binary label: 1 = Normal, 0 = any mental-health condition
    df1['label_binary'] = df1['label'].apply(
        lambda x: 1 if str(x).strip().lower() == 'normal' else 0
    )

    # Split Dataset 1 into Normal and non-Normal
    df1_normal     = df1[df1['label_binary'] == 1][['text', 'label_binary']].copy()
    df1_non_normal = df1[df1['label_binary'] == 0][['text', 'label_binary']].copy()

    print(f"\nDataset 1  — Normal rows    : {len(df1_normal)}")
    print(f"Dataset 1  — Non-Normal rows: {len(df1_non_normal)}")

    # ── Prepare Dataset 3 ─────────────────────────────────────────
    # Use 'content' as text; label is already 0 (depression/negative)
    df3 = df3.rename(columns={'content': 'text'})[['text', 'label']].copy()
    df3 = df3.rename(columns={'label': 'label_binary'})

    print(f"\nDataset 3  — all rows (label=0): {len(df3)}")

    # ── Build training set ────────────────────────────────────────
    # Dataset 3 (label=0)  +  Dataset 1 Normal rows (label=1)
    train_df = pd.concat([df3, df1_normal], ignore_index=True)
    train_df = train_df.dropna(subset=['text', 'label_binary'])
    train_df = train_df.sample(frac=1, random_state=42).reset_index(drop=True)  # shuffle

    print(f"\nTraining set size : {len(train_df)}")
    print("Training label distribution:")
    print(train_df['label_binary'].value_counts())

    # ── Build test set ────────────────────────────────────────────
    # Held-out non-Normal rows from Dataset 1 (all should be label=0)
    test_df = df1_non_normal.rename(columns={'label_binary': 'label_binary'}).copy()
    test_df = test_df.dropna(subset=['text', 'label_binary'])

    print(f"\nTest set size     : {len(test_df)}")
    print("Test label distribution:")
    print(test_df['label_binary'].value_counts())

    # ── Convert to lists / arrays ──────────────────────────────────
    train_texts  = train_df['text'].astype(str).tolist()
    test_texts   = test_df['text'].astype(str).tolist()
    train_labels = train_df['label_binary'].astype(int).values
    test_labels  = test_df['label_binary'].astype(int).values

    # ── Load tokenizer & model ────────────────────────────────────
    model_name = "distilbert-base-uncased-finetuned-sst-2-english"
    print(f"\nLoading model: {model_name}")
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model     = AutoModelForSequenceClassification.from_pretrained(
        model_name, num_labels=2
    )

    # ── Tokenise ──────────────────────────────────────────────────
    train_dataset = SentimentDataset(train_texts, train_labels, tokenizer)
    test_dataset  = SentimentDataset(test_texts,  test_labels,  tokenizer)

    # ── Training arguments ────────────────────────────────────────
    training_args = TrainingArguments(
        output_dir='./results',
        num_train_epochs=3,
        per_device_train_batch_size=16,
        per_device_eval_batch_size=16,
        warmup_steps=500,
        weight_decay=0.01,
        logging_dir='./logs',
        logging_steps=100,
        eval_strategy="epoch", # Changed from evaluation_strategy
        save_strategy="epoch",
        load_best_model_at_end=True,
    )

    # ── Trainer ───────────────────────────────────────────────────
    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=test_dataset,
        compute_metrics=compute_metrics,
    )

    print("\nStarting fine-tuning...")
    trainer.train()

    # ── Evaluate ──────────────────────────────────────────────────
    print("\nEvaluating on Dataset 1 (non-Normal rows)...")
    eval_results = trainer.evaluate()
    print(f"Evaluation Results: {eval_results}")

    # Full classification report
    predictions = trainer.predict(test_dataset)
    preds = np.argmax(predictions.predictions, axis=-1)

    print("\nClassification Report:")
    print(classification_report(
        test_labels, preds,
        target_names=["Depressed/Abnormal (0)", "Normal (1)"]
    ))

    # ── Save ──────────────────────────────────────────────────────
    save_path = './fine-tuned-d3-train-d1-test'
    print(f"\nSaving fine-tuned model to '{save_path}'...")
    model.save_pretrained(save_path)
    tokenizer.save_pretrained(save_path)
    print("Done!")


if __name__ == "__main__":
    main()
```
